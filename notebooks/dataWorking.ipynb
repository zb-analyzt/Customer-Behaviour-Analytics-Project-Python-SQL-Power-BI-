NOW WORK ON FILL SESSION DURATION AVERAGE

```python
df["session_duration_avg"].dtypes
```

```python
df["session_duration_avg"].isnull().mean()*100
```

```python
df["session_duration_avg"].median()
```

```python
df["membership_years_round"] = df["membership_years"].round()
```

```python
df.groupby("membership_years_round")["session_duration_avg"].median()
```

In this transform Means in each session duration lik 0-1,1-2 missing value replace with the specific years sesion

```python
df["session_duration_avg"]=df["session_duration_avg"].fillna(df.groupby
("membership_years_round")["session_duration_avg"].transform("median"))
```

IT is used as a Backup if still reamin any vcalue missing replace it With overall missing Values Median

```python
df["session_duration_avg"]=df["session_duration_avg"].fillna(df.groupby
("membership_years_round")["session_duration_avg"].median())
```
